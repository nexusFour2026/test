```python
def feature_engineering(data):
    data = data.copy()

    # --------------------------------------------------------
    # Mileage
    # Example: "81,000 mi." -> 81000
    # --------------------------------------------------------
    if "mileage" in data.columns:
        data["mileage"] = (
            data["mileage"]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("mi.", "", regex=False)
            .str.replace("mi", "", regex=False)
            .str.strip()
        )

        data["mileage"] = pd.to_numeric(
            data["mileage"],
            errors="coerce"
        )

```

```python
 if "engine" in data.columns:

        engine_text = data["engine"].astype(str)

        # Horsepower
        data["horsepower"] = pd.to_numeric(
            engine_text.str.extract(
                r"(\d+(?:\.\d+)?)\s*HP",
                expand=False
            ),
            errors="coerce"
        )

        # Engine liters
        data["engine_liters"] = pd.to_numeric(
            engine_text.str.extract(
                r"(\d+(?:\.\d+)?)\s*L",
                expand=False
            ),
            errors="coerce"
        )

        # Cylinder count
        data["cylinders"] = pd.to_numeric(
            engine_text.str.extract(
                r"(\d+)\s*(?:Cyl|Cylinder)",
                expand=False,
                flags=0
            ),
            errors="coerce"
        )
```
